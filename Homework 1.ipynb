Q7. Sum of weights

```python
asia = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
asia
```

```python
X = asia.to_numpy()
X
```

```python
XTX = X.T @ X
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
w = XTX_inv @ X.T @ y
w
```

```python
w.sum().round(3)
```
